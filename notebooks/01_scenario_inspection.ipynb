# SC06 — Scenario Inspection

This notebook inspects deterministic synthetic routing scenarios generated for algorithm testing. They are not real traffic observations.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data" / "processed").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
DATA_PATH = PROJECT_ROOT / "data" / "processed" / "sc06_route_scenarios.csv"
df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
print("Missing values:", int(df.isna().sum().sum()))
print("Duplicate rows:", int(df.duplicated().sum()))
print("Scenario counts:")
print(df["scenario_type"].value_counts().sort_index())
df.head()
```

## Validation notes

The generator validates scenario IDs, row count, numeric ranges, coordinates, traffic labels, and priorities before writing the CSV. Coordinates are schematic only; future real map/traffic data must be sourced and documented separately.
